# YOLOv8 training notebook

This notebook provides an example on how you can use Google Colab to train your groceries (fruits and veggies) detector with YOLOv8.

This notebook is a simplified version of the [tutorial](https://colab.research.google.com/github/roboflow-ai/notebooks/blob/main/notebooks/train-yolov8-object-detection-on-custom-dataset.ipynb) provided by Roboflow. Please feel free to check out their notebook for a more detailed tutorial.

```python
import json
import numpy as np

INPUT_PATH = '../lab_output/slam_state.json'
OUTPUT_PATH = '../lab_output/slam.txt'

with open(INPUT_PATH) as f:
    data = json.load(f)

taglist = data['taglist']
markers = np.array(data['markers'])  # shape (2, N): row0=x, row1=y

assert markers.shape[0] == 2, f"Expected markers shape (2, N), got {markers.shape}"
assert markers.shape[1] == len(taglist), "markers columns must match taglist length"

slam_dict = {}
for i, tag in enumerate(taglist):
    key = f"aruco{tag}_0"
    slam_dict[key] = {
        "x": float(markers[0, i]),
        "y": float(markers[1, i])
    }

with open(OUTPUT_PATH, 'w') as f:
    json.dump(slam_dict, f, indent=4)

print(f"Wrote {len(slam_dict)} markers to {OUTPUT_PATH}")
```
